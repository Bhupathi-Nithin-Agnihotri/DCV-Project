```python
oc = pd.read_csv("../clean/opencity_results_clean_2026-10-20.csv")
display(oc.sample(n=5))
cand = pd.read_csv("../interim/candidates_processed.csv")
display(cand.sample(n=5))
```

```python
cand['total_assets'] = pd.to_numeric(
    cand['total_assets'].str.extract(r'Rs\s*([\d,]+)\s*~')[0].str.replace(',', '', regex=False),
    errors='coerce'
)
cand['liabilities'] = pd.to_numeric(
    cand['liabilities'].str.extract(r'Rs\s*([\d,]+)\s*~')[0].str.replace(',', '', regex=False),
    errors='coerce'
)
```

```python
import pandas as pd
import re
from difflib import SequenceMatcher

def clean_name(name):
    if pd.isna(name): return ""
    # Remove titles
    name = re.sub(r'(?i)\b(dr|adv|prof|thiru|shri|smt|kum|ca|er|mr|mrs)\.?\b', '', str(name))
    # Remove special chars, lowercase, strip
    name = re.sub(r'[^a-zA-Z\s]', '', name).lower().strip()
    # Sort words alphabetically
    return ' '.join(sorted(name.split()))

# Standardize state and constituency (removes spaces/special chars for reliable merging)
cand['match_state'] = cand['state'].str.lower().str.replace(r'[^a-z]', '', regex=True)
cand['match_const'] = cand['const'].str.lower().str.replace(r'[^a-z]', '', regex=True)
cand['clean_name'] = cand['candidate'].apply(clean_name)

oc['match_state'] = oc['State'].str.lower().str.replace(r'[^a-z]', '', regex=True)
oc['match_const'] = oc['Constituency_Name'].str.lower().str.replace(r'[^a-z]', '', regex=True)
oc['clean_name'] = oc['Candidate'].apply(clean_name)
```
