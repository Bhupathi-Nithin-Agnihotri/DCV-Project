**Collection Method:** csv files downloaded programmatically using Python's `requests` library. The notebook skips downloading a file if it already exists locally.

**Purpose:** obtain candidate-level election results, including votes and constituency information, to identify winners and losers

**Note:** the historical CSV contains election records from multiple years, including 2014 and 2019!! The year-specific datasets are extracted from this file.

### OpenCity robots.txt

**URL:** https://data.opencity.in/robots.txt

![OpenCity robots.txt rules](image.png)

**Findings:**
- `User-agent: *` means the rules apply to all crawlers.
- Dataset rating pages, revision pages, dataset history pages, and API endpoints are disallowed.
- `Crawl-Delay: 10` specifies a 10-second delay between crawler requests.

The CSV download URLs used in this project do not appear to match the listed disallowed paths.

```python
# required imports
import requests
import pandas as pd
import os
import time
from pathlib import Path
```

```python
# initialize session
session = requests.Session()

# set the user agent
session.headers.update({
    'User-Agent': "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537 .36 Chrome/"
})

# dataset urls
URL_2024 = "https://data.opencity.in/dataset/85a345c6-78c0-4f57-adfc-236c726c5456/resource/d164b73a-b855-4b68-be0c-0f3450e7ab9f/download/1b837c18-4f7a-4acb-aad0-918c51186a54.csv"
URL_2019 = "https://data.opencity.in/dataset/1874a140-a8c4-4d5d-80cd-cbbd400c21e5/resource/ed40f56f-3d97-41a4-8b5d-14b204b6f320/download/81842470-a23e-4785-a22b-ba81284101de.csv"
```

```python
# download the datasets if you haven't
datasets = {
    "2024": (URL_2024, "../../raw/opencity_data_2024.csv"),
    "2019": (URL_2019, "../../raw/opencity_data_2019.csv")
}

for year, (url, csv_path) in datasets.items():
    if not os.path.exists(csv_path):
        # download the csv file
        response = session.get(url, timeout=10)
        response.raise_for_status()

        # save the downloaded file
        with open(csv_path, "wb") as file:
            file.write(response.content)

        print(f"{year} dataset downloaded successfully!!")

        # delay
        time.sleep(10)
    else:
        print(f"{year} dataset already exists, skipping download")
```

```python
# load 2024 data
df_2024 = pd.read_csv("../../raw/opencity_data_2024.csv")

# light data inspection
print("First 5 rows:\n", df_2024.head())
print("\nShape:", df_2024.shape)
print("\nColumns:", df_2024.columns)
```

```python
# load 2019 data 
df_2019 = pd.read_csv("../../raw/opencity_data_2019.csv")

# light data inspection
print("First 5 rows:\n", df_2019.head())
print("\nShape:", df_2019.shape)
print("\nColumns:", df_2019.columns)
```

```python
# further inspect the 2019 dataset since there are many more rows compared to the 2024 dataset
print("Election years:\n", df_2019["Year"].value_counts().sort_index())
print("\nElection types:\n", df_2019["Election_Type"].value_counts())
```

```python
# turns out df_2019 contains multiple years, including 2014 and 2019
# preserve the full dataset under a more appropriate name
df_historical = df_2019.copy()

# extract seperate datasets for the 2014 and 2019 elections
df_2014 = df_historical[df_historical["Year"] == 2014].copy()
df_2019 = df_historical[df_historical["Year"] == 2019].copy()
```

```python
# light inspection for 2014 data
print("First 5 rows:\n", df_2014.head())
print("\nShape:", df_2014.shape)
print("\nColumns:", df_2014.columns)
```

```python
# light inspection for 2019 data
print("First 5 rows:\n", df_2019.head())
print("\nShape:", df_2019.shape)
print("\nColumns:", df_2019.columns)
```

```python
interim_dir = "../../interim"

for year, df in [(2014, df_2014), (2019, df_2019), (2024, df_2024)]:
    df.to_csv(f"{interim_dir}/opencity_electiondata_{year}.csv", index=False)

print("year-wise data sets saved successfully!")
```
