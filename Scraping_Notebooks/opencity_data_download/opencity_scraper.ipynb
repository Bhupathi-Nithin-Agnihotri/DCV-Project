```python
# required imports
import requests
import pandas as pd
import os
import time
```

```python
# initialize session
session = requests.Session()

# set the user agent
session.headers.update({
    'User-Agent': "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537 .36 Chrome/"
})

# dataset urls
URL_2024 = "https://data.opencity.in/dataset/85a345c6-78c0-4f57-adfc-236c726c5456/resource/d164b73a-b855-4b68-be0c-0f3450e7ab9f/download/1b837c18-4f7a-4acb-aad0-918c51186a54.csv"
URL_2019 = "https://data.opencity.in/dataset/1874a140-a8c4-4d5d-80cd-cbbd400c21e5/resource/ed40f56f-3d97-41a4-8b5d-14b204b6f320/download/81842470-a23e-4785-a22b-ba81284101de.csv"
```

```python
# download the datasets if you haven't
datasets = {
    "2024": (URL_2024, "opencity_data_2024.csv"),
    "2019": (URL_2019, "opencity_data_2019.csv")
}

for year, (url, csv_path) in datasets.items():
    if not os.path.exists(csv_path):
        # delay
        time.sleep(10)

        # download the csv file
        response = session.get(url, timeout=10)
        response.raise_for_status()

        # save the downloaded file
        with open(csv_path, "wb") as file:
            file.write(response.content)

        print(f"{year} dataset downloaded successfully")
    else:
        print(f"{year} dataset already exists, skipping download")
```

### Purpose of the Kept Columns

We kept only the columns needed for our project of comparing candidates' declared assets across the 2014, 2019 and 2024 Lok Sabha elections.

- **Year and constituency details:** Identify when and where each candidate contested.
- **Candidate and party:** Identify candidates and help match them with MyNeta's asset records.
- **Position:** Identify winners in the 2014 and 2019 elections.
- **Votes, vote share and margin:** Compare candidates' election performance.
- **Total Votes (2024):** Identify the winner in each constituency by comparing candidates' votes.

The remaining columns were excluded because they are not directly relevant for our problem. 
The original CSV files remain unchanged.

```python
# keep relevant historical columns
historical_cols = [
    "Year", "State_Name", "Constituency_No", "Constituency_Name",
    "Candidate", "Party", "Position", "Votes",
    "Vote_Share_Percentage", "Margin"
]

df_2014 = df_2014[historical_cols].copy()
df_2019 = df_2019[historical_cols].copy()

# keep relevant 2024 columns
cols_2024 = [
    "State", "PC No", "PC Name", "Candidate", "Party",
    "EVM Votes", "Postal Votes", "Total Votes"
]

df_2024 = df_2024[cols_2024].copy()
```

```python
# verify the final datasets
datasets = {"2014": df_2014, "2019": df_2019, "2024": df_2024}

for year, df in datasets.items():
    print(f"{year}: {df.shape[0]} rows, {df.shape[1]} columns")

print("notebook runs succesfully!!")
```
