```python
# required imports
import os
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
from urllib.parse import urljoin
```

```python
# initialize session
session = requests.Session()

# set the user agent
session.headers.update({
    'User-Agent': "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537 .36 Chrome/"
})

# the page url
URL = "https://data.opencity.in/dataset/parliamentary-elections-2024-results"

# fetch the first page
response = session.get(URL, timeout=10)

# catch errors
response.raise_for_status()
print(f'stats code: {response.status_code}')
```

```python
# parse the html
soup = BeautifulSoup(response.content, "html.parser")

# check if it worked fine
print(soup.title.get_text(strip=True))
```

```python
#soup
```

```python
# maps titles to data-set links
data_sets = {} 

for resource in soup.select("li.resource-item"):
    title = resource.select_one("a.heading").get_text(strip=True)
    link = resource.select_one("a.resource-url-analytics")
    url = link["href"]

    data_sets[title] = url

    print(title)
    print(url)
    # blank for better formatting
    print()  
```

```python
# lok sabha results are at this url
lsr_url = data_sets["Lok Sabha Elections 2024 ResultsCSV"]

csv_path = "opencity_data.csv"

# download the dataset if you already haven't
if not os.path.exists(csv_path):
    csv_response = session.get(lsr_url, timeout=30)
    csv_response.raise_for_status()

    with open(csv_path, "wb") as file:
        file.write(csv_response.content)

    print("Dataset downloaded successfully!")

else:
    print("Dataset already exists. Skipping download")
```

```python
# load the downloaded dataset
csv_path = "opencity_data.csv"
df = pd.read_csv(csv_path)

#inspect the dataset
print(df.head())
print(df.shape)
print(df.columns)
```

```python
# check missing values
print(df.isnull().sum())
print()

# check duplicate rows
print(f'Duplicate rows:', df.duplicated().sum())
print()

# check data types
print(df.dtypes)
print()
```
